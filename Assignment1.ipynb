```python
#4. Write a program to  check if a given word is a Python keyword.
import keyword

def is_python_keyword(word):
    """
    Checks if the given word is a Python keyword.

    Args:
        word (str): The word to check.

    Returns:
        bool: True if the word is a Python keyword, False otherwise.
    """
    return keyword.iskeyword(word)

# Example usage
if __name__ == "__main__":
    word = input("Enter a word to check if it is a Python keyword: ").strip()
    if is_python_keyword(word):
        print(f"'{word}' is a Python keyword.")
    else:
        print(f"'{word}' is not a Python keyword.")
```

```python
#9 Write a programm to convert user input from string to integer,float and boolean types.
def convert_input(user_input):
    results = {}

    try:
        results['integer'] = int(user_input)
    except ValueError:
        results['integer'] = "Cannot convert to integer"


    try:
        results['float'] = float(user_input)
    except ValueError:
        results['float'] = "Cannot convert to float"


    if user_input.lower() in ['true', 'false']:
        results['boolean'] = user_input.lower() == 'true'
    else:
        results['boolean'] = "Cannot convert to boolean"

    return results



if __name__ == "__main__":
    user_input = input("Enter a value: ").strip()

    conversions = convert_input(user_input)

    print("\nConversions:")
    for key, value in conversions.items():
        print(f"As {key}: {value}")
```

```python
#10 Write code to demonstrate type casting with list elements.
def demonstrate_type_casting(input_list):
    print("Original List:", input_list)

    converted_list = []
    for item in input_list:
        try:

            converted_list.append(int(item))
            print(f"Converted '{item}' to Integer.")
        except ValueError:
            try:

                converted_list.append(float(item))
                print(f"Converted '{item}' to Float.")
            except ValueError:
                if item.lower() in ['true', 'false']:

                    converted_list.append(item.lower() == 'true')
                    print(f"Converted '{item}' to Boolean.")
                else:

                    converted_list.append(item)
                    print(f"Kept '{item}' as String.")

    return converted_list



if __name__ == "__main__":

    input_list = ["123", "45.67", "True", "False", "hello", "42"]

    print("\nDemonstrating Type Casting:")
    result = demonstrate_type_casting(input_list)

    print("\nConverted List:", result)
```

```python
#14 Write a program to reverse a string a string using while loop.
def reverse_string_while_loop(input_string):
    reversed_string = ""
    index = len(input_string) - 1

    while index >= 0:
        reversed_string += input_string[index]
        index -= 1
    return reversed_string



if __name__ == "__main__":

    user_input = input("Enter a string to reverse: ")

    reversed_result = reverse_string_while_loop(user_input)


    print("Reversed String:", reversed_result)
```

```python
#15 Write a python programto calculate teh factorial of a number provided by the userusing a while loop.
def calculate_factorial(number):
    if number < 0:
        return "Factorial is not defined for negative numbers."

    factorial = 1
    while number > 0:
        factorial *= number
        number -= 1

    return factorial



if __name__ == "__main__":
    try:

        user_input = int(input("Enter a non-negative integer: "))


        result = calculate_factorial(user_input)


        print(f"The factorial of {user_input} is: {result}")
    except ValueError:
        print("Invalid input! Please enter a valid non-negative integer.")
```
